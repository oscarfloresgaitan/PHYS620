#### Mass on the surface of a cone (Taylor Example 13.4)

Consider a mass $m$ constrained to move on the frictionless surface of a vertical cone $\rho = cz$ (in cylindrical polar coordinates $\rho, \phi, z$ with $z > 0$) in a uniform gravitational field $g$ vertically downward.

From Taylor Example 13.4, the kinetic energy is:
$$T = \frac{1}{2}m\left[(c^2+1)\dot{z}^2 + (cz\dot{\phi})^2\right]$$

and the potential energy is $U = mgz$. The Lagrangian is $\mathcal{L} = T - U$.

The generalized momenta are:
$$p_z = \frac{\partial T}{\partial \dot{z}} = m(c^2+1)\dot{z} \quad \text{and} \quad p_\phi = \frac{\partial T}{\partial \dot{\phi}} = mc^2z^2\dot{\phi}$$

The Hamiltonian is:
$$\mathcal{H} = T + U = \frac{1}{2m}\left[\frac{p_z^2}{c^2+1} + \frac{p_\phi^2}{c^2z^2}\right] + mgz$$

Since $\dot{z} = 0$ at $z_{\text{min}}$ and $z_{\text{max}}$, the vertical momentum $p_z = 0$. Setting $p_z = 0$ gives:
$$\mathcal{H}(z) = \frac{p_\phi^2}{2mc^2z^2} + mgz$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
m = 1.0       # mass (kg)
g = 9.8       # gravity (m/s^2)
c = 1.0       # constant
p_phi = 3.0   # phi

# Height
z = np.linspace(0.25, 3.0, 300)

# Hamiltonian with p_z = 0
H = (p_phi**2) / (2 * m * c**2 * z**2) + m * g * z

# Total energy
E = 20.0

# Plot
plt.figure(figsize=(7, 4.5))
plt.plot(z, H, label=r'$\mathcal{H}(z)$ with $p_z = 0$', color='blue', lw=2)
plt.axhline(E, color='red', linestyle='--', label='Energy $E$')

plt.xlabel('Height $z$ (m)')
plt.ylabel(r'$\mathcal{H}(z)$ (J)')
plt.title(r'Mass on a Cone')
plt.ylim(0, 40)
plt.xlim(0.2, 3.0)
plt.grid(True)
plt.legend()
plt.show()
```

#### Description of the Motion

- The Hamiltonian $\mathcal{H}$ equals the conserved total energy $E$. Setting $p_z = 0$ (since $\dot{z} = 0$ at both extrema) gives the curve $\mathcal{H}(z)$. The horizontal line $E$ intersects this curve at $z_{\text{min}}$ and $z_{\text{max}}$. Because $E \ge \mathcal{H}(z)$, the mass is confined to oscillate up and down between $z_{\text{min}}$ and $z_{\text{max}}$.

- As $z \to 0$, the centrifugal term $\frac{p_\phi^2}{2mc^2z^2} \to \infty$. This prevents the mass from ever falling all the way to the bottom of the cone at $z = 0$.
- As $z \to \infty$, the term $mgz \to \infty$. Since $\mathcal{H} = E$ is fixed, there is an upper bound for $z_{\text{max}}$.
