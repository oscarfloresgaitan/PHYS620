#### Coriolis force

When a puck slides on a rotating turntable, it can come instantaneously to rest. From a perspective looking straight down on the turntable, plot the shape of the puck's path before it comes to rest. In a markdown cell, explain why this happens.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
b = 1.0
omega = 1.0

# Time before coming to rest
t = np.linspace(-3 * np.pi, 0, 500)

# Trajectory in the rotating frame
x = b * (np.cos(omega * t) + omega * t * np.sin(omega * t))
y = b * (-np.sin(omega * t) + omega * t * np.cos(omega * t))

# Plot
plt.figure(figsize=(6, 6))
plt.plot(x, y, color='navy', label='Puck path')
plt.plot(x[-1], y[-1], 'ro', label='Instantaneous rest')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Puck Path on Rotating Turntable')
plt.axis('equal')
plt.grid(True)
plt.legend()
plt.show()
```

The puck's velocity in the turntable frame is $\vec{v}_{\mathrm{rot}} = \vec{v}_0 - \vec{\Omega} \times \vec{r}$. Along its straight inertial path, the puck passes at distance $b = v_0 / \Omega$ from the rotation axis. At this point of closest approach, the turntable surface moves at speed $\Omega b = v_0$ in the direction of $\vec{v}_0$. The relative velocity drops to zero, and the puck comes instantaneously to rest.

#### Surface deformation due to spinning

A bucket of water is spun about its vertical axis with angular speed $\Omega$. The water in the very center of the bucket has depth $h_0$. Assume the water has settled in equilibrium relative to the bucket. Plot the height of the water surface $h - h_0$ across a diameter of the bucket. To find $h(r)$, use cylindrical polar coordinates and remember that the surface is an equipotential under the combined effects of the gravitational and centrifugal forces.

```python
# Parameters
g = 9.8               # (m/s^2)
R_bucket = 0.2        # radius (m)
Omega = 10.0           # angular velocity (rad/s)

# Radial coordinate 
r = np.linspace(-R_bucket, R_bucket, 200)

# Surface height relative to center
dh = (Omega**2 / (2 * g)) * (r**2)

# Plot
plt.figure(figsize=(7, 4.5))
plt.plot(r, dh, color='teal')
plt.xlabel('Radial position (m)')
plt.ylabel('Height (m)')
plt.grid(True)
plt.show()
```

#### Ball drop on rotating Earth

A ball is dropped from a very tall stepladder on the equator. Plot the ball's path $z(x)$ from a tower due north of the stepladder. Explain in a markdown cell why the object lands east of its point of release.

```python
# Parameters
g = 9.8                         # (m/s^2)
h = 100.0                       # drop height (m)
Omega_E = 7.292e-5              # Earth's angular speed (rad/s)

# Total drop time
t_drop = np.sqrt(2 * h / g)
t = np.linspace(0, t_drop, 500)

x = (1.0 / 3.0) * Omega_E * g * (t**3) # Horizontal deflection x(t)
z = h - 0.5 * g * (t**2)               # Height z(t)

# Plot
plt.figure(figsize=(7, 5))
plt.plot(x, z, color='darkred')
plt.xlabel('Horizontal displacement $x$ (m)')
plt.ylabel('Height $z$ (m)')
plt.title('Ball Drop Trajectory on Rotating Earth')
plt.grid(True)
plt.show()
```

The ball lands east of its release point for two reasons:

1. **Inertial frame:** Prior to release, the ball at height $h$ sits at distance $r = R_E + h$ from Earth's rotation axis, so it has an initial eastward velocity $v_{\mathrm{east}} = \Omega(R_E + h)$. The ground moves at $v_{\mathrm{surface}} = \Omega R_E$.

2. **Rotating frame:** As the ball falls downward with velocity $\vec{v} \approx -gt\,\hat{z}$, it experiences a Coriolis acceleration directed toward the east:
$$\vec{a}_{\mathrm{Cor}} = -2(\vec{\Omega} \times \vec{v}) = -2(\Omega\,\hat{y} \times (-gt\,\hat{z})) = 2\Omega gt\,\hat{x}$$
